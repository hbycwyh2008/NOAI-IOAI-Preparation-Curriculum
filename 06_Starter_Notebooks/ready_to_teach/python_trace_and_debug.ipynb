## Entry check

2. State one loop invariant.
3. Give one boundary case and one malformed-input case.


```python
values = [3, -1, 4, 0, 2]
total = count = 0
trace = []
for value in values:
    if value > 0:
        total += value
        count += 1
    trace.append((value, total, count))
print(trace)
assert total == 9 and count == 3

def safe_average(values):
    cleaned = [float(v) for v in values if v is not None]
    if not cleaned:
        raise ValueError('no usable values')
    return sum(cleaned) / len(cleaned)
assert safe_average([1, 2, None, 3]) == 2.0
```
